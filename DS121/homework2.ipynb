```python
# Import necessary libraries
%pip install numpy pandas scikit-learn
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
```

The Caravan_Insurance dataset includes 85 independent variables (features) that measure demographic characteristics for 5,822 individuals. The dependent (or target) variable is Purchase, which indicates whether or not a given individual purchases a caravan insurance policy. 

### Part a

```python
# Load the dataset from the Caravan_Insurance.csv. 
Caravan = pd.read_csv("Caravan_Insurance.csv")
Caravan.head()
```

The cell below computes the percentage of people who purchased a caravan insurance policy.

### Part b

```python
# Extract the target variable (whether a customer made a purchase)
Purchase = Caravan["Purchase"]

# Display the number of "Yes" and "No" responses in the target variable
print(Purchase.value_counts())

# Calculate and print the percentage of customers who made a purchase ("Yes")
percentage_yes = (Purchase.value_counts()["Yes"] / Purchase.count()) * 100
print(f"Percentage of purchases: {percentage_yes:.2f}%")
```

Because the KNN classifier predicts the class of a given test data point by identifying the data points that are nearest to it, the scale of the variables matters. Any variables that are on a large scale will have a much larger effect on the distance between the data sets, and hence on the KNN classifier, than variables that are on a small scale.

A good way to handle this problem is to standardize the data so that all standardize variables are given a mean of zero and a standard deviation of one. Then all variables will be on a comparable scale. This is accomplished using the StandardScaler() transformation. This is what is done in the cell below.

### Part c

```python
# Separate the feature columns (drop the target variable)
feature_df = Caravan.drop(columns=["Purchase"])

# Standardize the feature values to have mean 0 and standard deviation 1
scaler = StandardScaler()
X_std = scaler.fit_transform(feature_df)
```

Using the function train_test_split() we now split the observations into a test set, containing 1000 observations, and a training set containing the train_test_split() remaining observations. The argument random_state=0 ensures that we get the same split each time we rerun the code.

### Part d

```python
# Split the data into training and test sets
# Use 1000 samples for testing, the rest for training
X_train, X_test, y_train, y_test = train_test_split(X_std, Purchase, test_size=1000, random_state=0)
```

We fit a $k$NN model on the training data using $k = 1$, and evaluate its performance on the test data.

### Part e

```python
# Train a 1-nearest neighbor classifier and make predictions on the test set
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1_pred = knn1.fit(X_train, y_train).predict(X_test)

# Compute and print the confusion matrix for k=1
cm = confusion_matrix(y_test, knn1_pred, labels=["Yes", "No"])
print("Confusion Matrix:\n", cm)
```

### Part f

```python
# Evaluate KNN classifiers for K nearest neighbors
for K in range(2, 6):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_matrix(y_test, knn_pred, labels=["Yes", "No"])
    
    # Extract values from the confusion matrix
    predicted_yes = C[:, 0].sum()  # All predicted as 'Yes'
    actual_yes_predicted_yes = C[0, 0]  # True positives
    
    accuracy = actual_yes_predicted_yes / predicted_yes if predicted_yes != 0 else 0
    print(f"K={K}: # predicted to purchase: {predicted_yes:>2}, # who did purchase: {actual_yes_predicted_yes}, accuracy: {accuracy:.1%}")
```
